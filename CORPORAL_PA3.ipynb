## POSITIONAL AND LABEL-BASED SLICING

```python
import pandas as pd
cars = pd.read_csv('cars.csv')
cars

print("Shape of the dataframe cars: ", cars.shape)
print("\nThe column names of cars: \n\n", cars.columns.tolist(), "\n")

cars_6_to_10 = cars.iloc[5:10]
cars_6_to_10.loc[:, ['Model', 'mpg', 'cyl', 'hp', 'gear']]
```

##  MODEL LOOKUP

```python
import pandas as pd
cars=pd.read_csv('cars.csv')
cars

toyota = cars[cars['Model'] == 'Toyota Corolla']
pontiac = cars[cars['Model'] == 'Pontiac Firebird']

display(toyota)
display(pontiac.loc[:, ['Model', 'mpg', 'hp', 'wt']])
```

## MULTI-MODEL SUBSETTING

```python
import pandas as pd
cars=pd.read_csv('cars.csv')
cars

selected_cars = cars[(cars['Model'] == 'Datsun 710') | (cars['Model'] == 'Lotus Europa') | (cars['Model'] == 'Ferrari Dino')]
selected_cars = selected_cars.loc[:, ['Model', 'mpg', 'cyl', 'hp', 'gear']]

print("The shape of the selected cars is: ", selected_cars.shape, "\n")
display(selected_cars)
```
